#**4. Pre Processing**

```python
data_processed = df.copy()
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
data_processed['Status Gizi'] = label_encoder.fit_transform(data_processed['Status Gizi'])
```

```python
X = data_processed.drop('Status Gizi', axis=1)
y = data_processed['Status Gizi']
```

```python
# 3. One-Hot Encoding untuk Fitur Kategorikal ('Jenis Kelamin')
# Mengubah 'Jenis Kelamin' menjadi kolom numerik tanpa membuat urutan palsu
X = pd.get_dummies(X, columns=['Jenis Kelamin'], drop_first=True)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
# 4. Scaling Fitur
# Menyamakan skala semua fitur agar tidak ada fitur yang mendominasi
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42, stratify=y)

print("Dimensi data latih (X_train):", X_train.shape)
print("Dimensi data uji (X_test):", X_test.shape)
```

#**5. Model KNN**

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
# Mencari nilai k terbaik dengan "Elbow Method"
error_rate = []

# Kita akan coba nilai k dari 1 hingga 10
for i in range(1, 11):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train, y_train)
    pred_i = knn.predict(X_test)
    error_rate.append(np.mean(pred_i != y_test))

# Membuat plot untuk visualisasi hasil
plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), error_rate, color='blue', linestyle='dashed', marker='o',
         markerfacecolor='red', markersize=10)
plt.title('Error Rate vs. Nilai K')
plt.xlabel('Nilai K')
plt.ylabel('Error Rate')
plt.grid(True)
plt.show()

# Melatih model final dengan k optimal (misalnya, k=5 berdasarkan pengamatan)
k_optimal = 5
model_knn = KNeighborsClassifier(n_neighbors=k_optimal)
model_knn.fit(X_train, y_train)

print(f"Model KNN telah dilatih dengan k = {k_optimal}.")
```

#**6. Evaluasi KNN**

```python
# Membuat prediksi pada data uji
y_pred = model_knn.predict(X_test)

# 1. Menghitung Akurasi
accuracy = accuracy_score(y_test, y_pred)
print(f"Akurasi Model: {accuracy:.4f}")

# 2. Menampilkan Classification Report
# Laporan ini memberikan rincian presisi, recall, dan f1-score per kelas
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=label_encoder.classes_))

# 3. Visualisasi Confusion Matrix
# Matrix ini menunjukkan berapa banyak prediksi yang benar dan salah untuk setiap kelas
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=label_encoder.classes_,
            yticklabels=label_encoder.classes_)
plt.title('Confusion Matrix')
plt.ylabel('Label Aktual')
plt.xlabel('Label Prediksi')
plt.show()
```

```python
# Impor library yang dibutuhkan
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.cluster import KMeans
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder

# --- 1. Memuat Dataset ---
# Memuat dataset data_balita.csv
df = pd.read_csv('data_balita.csv')

# --- 2. Analisis Data Eksploratif (EDA) ---
print("--- Analisis Data Eksploratif (EDA) ---")
print("Lima baris pertama dari dataset:")
print(df.head())
print("\nInformasi dataset:")
df.info()
print("\nStatistik deskriptif:")
print(df.describe())
print("\nDistribusi kelas Status Gizi:")
print(df['Status Gizi'].value_counts())

# Visualisasi distribusi Status Gizi
plt.figure(figsize=(8, 6))
sns.countplot(x='Status Gizi', data=df)
plt.title('Distribusi Status Gizi Balita')
plt.show()

# --- 3. Pra-pemrosesan Data ---
print("\n--- Pra-pemrosesan Data ---")
# Membuat salinan dataframe untuk menghindari perubahan pada data asli
df_processed = df.copy()

# Mengubah data kategorikal menjadi numerik menggunakan LabelEncoder
label_encoder = LabelEncoder()
df_processed['Jenis Kelamin'] = label_encoder.fit_transform(df_processed['Jenis Kelamin'])
# 'Gizi Baik': 0, 'Gizi Buruk': 1, 'Gizi Lebih': 2, 'Stunting': 3
df_processed['Status Gizi'] = label_encoder.fit_transform(df_processed['Status Gizi'])

print("Data setelah encoding:")
print(df_processed.head())

# --- 4. Analisis Clustering (K-Means) ---
print("\n--- Analisis Clustering (K-Means) ---")
# Memilih fitur untuk clustering (Umur dan Tinggi Badan)
X_cluster = df_processed[['Umur (bulan)', 'Tinggi Badan (cm)']]

# Menentukan jumlah cluster yang optimal menggunakan metode Elbow
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(X_cluster)
    wcss.append(kmeans.inertia_)

# Plot metode Elbow
plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.title('Metode Elbow untuk Menentukan Jumlah Cluster Optimal')
plt.xlabel('Jumlah Cluster')
plt.ylabel('WCSS')
plt.show()

# Berdasarkan metode Elbow, kita memilih 4 cluster (sesuai jumlah kategori status gizi)
kmeans = KMeans(n_clusters=4, init='k-means++', random_state=42, n_init=10)
y_kmeans = kmeans.fit_predict(X_cluster)
df_processed['Cluster'] = y_kmeans

# Visualisasi hasil clustering
plt.figure(figsize=(12, 8))
sns.scatterplot(x='Umur (bulan)', y='Tinggi Badan (cm)', hue='Cluster', data=df_processed, palette='viridis', s=100)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=300, c='red', label='Centroids', marker='*')
plt.title('Hasil Clustering K-Means Data Balita')
plt.xlabel('Umur (bulan)')
plt.ylabel('Tinggi Badan (cm)')
plt.legend()
plt.show()

print("Hasil clustering:")
print(df_processed[['Umur (bulan)', 'Tinggi Badan (cm)', 'Cluster']].head())

# --- 5. Analisis Klasifikasi (Decision Tree) ---
print("\n--- Analisis Klasifikasi (Decision Tree) ---")
# Memilih fitur (X) dan target (y)
X_class = df_processed[['Umur (bulan)', 'Jenis Kelamin', 'Tinggi Badan (cm)']]
y_class = df_processed['Status Gizi']

# Membagi data menjadi data latih dan data uji
X_train, X_test, y_train, y_test = train_test_split(X_class, y_class, test_size=0.3, random_state=42)

# Membuat dan melatih model Decision Tree
dt_classifier = DecisionTreeClassifier(random_state=42)
dt_classifier.fit(X_train, y_train)

# Memprediksi data uji
y_pred = dt_classifier.predict(X_test)

# Evaluasi model
accuracy = accuracy_score(y_test, y_pred)
print(f"Akurasi Model Decision Tree: {accuracy:.2f}")
print("\nLaporan Klasifikasi:")
print(classification_report(y_test, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# --- 6. Analisis Aturan Asosiasi (Apriori) ---
print("\n--- Analisis Aturan Asosiasi (Apriori) ---")
# Untuk aturan asosiasi, kita perlu data dalam format transaksional
# Kita akan mengubah data menjadi format biner (one-hot encoding)
# Kita gunakan data asli sebelum di-encode
df_assoc = df.copy()

# Mengubah 'Status Gizi' menjadi item untuk asosiasi
df_assoc['Umur (kategori)'] = pd.cut(df_assoc['Umur (bulan)'], bins=[0, 12, 24, 36, 48, 60], labels=['0-1 thn', '1-2 thn', '2-3 thn', '3-4 thn', '4-5 thn'])
df_assoc_onehot = pd.get_dummies(df_assoc[['Jenis Kelamin', 'Status Gizi', 'Umur (kategori)']], prefix='', prefix_sep='')

# Menjalankan algoritma Apriori
frequent_itemsets = apriori(df_assoc_onehot, min_support=0.1, use_colnames=True)
print("Frequent Itemsets (min_support=0.1):")
print(frequent_itemsets)

# Membuat aturan asosiasi
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
print("\nAturan Asosiasi (min_lift=1):")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].sort_values(by='lift', ascending=False))
```
